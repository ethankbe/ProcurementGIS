```python
import pandas as pd
```

```python
df2 = pd.read_csv("philgeps2.csv", encoding="latin-1", low_memory=False)
print(df2.shape)
df2.columns = df2.columns.str.strip()
```

```python
cols2 = [
    "Procuring Entity (PE)", "Region", "Province", "City/Municipality",
    "Bid Reference No.", "Notice Title", "Classification", "Procurement Category",
    "Approved Budget for the Contract (ABC)", "Published Date", "Closing Date",
    "Item/Lot Name", "Item/Lot Description", "Quantity", "Unit of Measure", "Item Budget", "Bid Notice Status", "Award Reference No."
]
df2 = df2[cols2].copy()
```

```python
# change philgeps2.csv to file name

awards = pd.read_csv("philgeps2.csv", encoding="latin-1", usecols=["Procurement Category"], low_memory=False)

cats = sorted(awards["Procurement Category"].dropna().str.strip().unique())

with open("procurement_categories_philgeps2.txt", "w", encoding="utf-8") as f:
    for c in cats:
        f.write(c + "\n")

print(len(cats), "unique categories saved")
```

```python
# 1. NCR / Metro Manila
ncr2 = df2[
    (df2["Region"].str.upper() == "NCR") &
    (df2["Province"].str.lower() == "metro manila")
].copy()

# 2. IT categories
it_cats2 = [
    "Computer Equipment and Accessories",
    "Computer services",
    "Software",
    "Information Technology Service Delivery",
    "Information services",
    "Electronic reference material",
    "Data Voice or Multimedia Network Equipment or Platforms and Accessories",
    "Components for information technology or broadcasting or telecommunications",
    "Communications Devices and Accessories",
    "Telecommunications media services",
    "Electronic hardware and component parts and accessories",
    "Printed circuits and integrated circuits and microassemblies",
    "Office machines and their supplies and accessories",
    "Audio and visual presentation and composing equipment",
    "Photographic or filming or video equipment",
    "Photographic and recording media",
    "Consumer electronics",
    "Security surveillance and detection",
    "Automation control devices and components and accessories",
    "Electrical wire and cable and harness",
    "Electrical wire management devices and accessories and supplies",
    "Batteries and generators and kinetic power transmission",
    "Printing and publishing equipment",
]

# 3. Awarded only
ncr_it2 = ncr2[
    ncr2["Procurement Category"].isin(it_cats2) &
    (ncr2["Bid Notice Status"].str.lower() == "awarded")
].copy()
```

```python
for c in ["Approved Budget for the Contract (ABC)", "Item Budget"]:
    ncr_it2[c] = pd.to_numeric(
        ncr_it2[c].astype(str).str.replace(r"[^\d.\-]", "", regex=True),
        errors="coerce"
    )

ncr_it2["Published Date"] = pd.to_datetime(
    ncr_it2["Published Date"], dayfirst=True, errors="coerce"
)

ncr_it2.to_csv("ncr_it_awarded2.csv", index=False, encoding="utf-8-sig")
```
